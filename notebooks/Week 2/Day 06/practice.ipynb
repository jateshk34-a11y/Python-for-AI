```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if arr[mid] == target:
            return mid, comparisons

        elif target < arr[mid]:
            high = mid - 1

        else:
            low = mid + 1

    return -1, comparisons


# Student roll numbers
students = [101, 102, 103, 104, 105, 106, 107, 108]

target = 105

index, comparisons = binary_search(students, target)

print("Index:", index)
print("Comparisons:", comparisons)
```
