**9.** Given `["Ali", "Fatima", "Hassan"]` and `[88, 72, 95]`, print `1: Ali - 88` for each pair, numbered from 1. Use `enumerate()` and `zip()` together.

````python
names = ["Ali", "Fatima", "Hassan"]
scores = [88, 72, 95]

for i, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{i}: {name} - {score}")
````

Task: Write a Python program that:
1. Creates a list containing "Coffee", "Tea", "Juice", and "Water".
2. Uses enumerate() to display each item with a number starting from 1.
3. Prints the menu in this format:
1 Coffee
2 Tea
3 Juice
4 Water

````python
lists = ["Coffee","Tea","Juice","Water"]
for start, list in enumerate(lists, start=1):
    print(start,list)
````

Task: Write a Python program that:
1. Creates a list of student names: ["Ali", "Fatima", "Hassan"].
2. Creates a list of marks: [85, 92, 78].
3. Uses zip() to loop through both lists together.
4. Prints each student's name with their marks in this format:

````python
stu_name = ["ali","fatima","hassan"]
marks = [85,92,78]
for stu_name, mark in zip(stu_name,marks):
    print(stu_name,"scored",mark)
````

Task: Write a Python program that:
1. Creates a list of numbers from 1 to 20.
2. Uses list comprehension to create a new list containing only even numbers.
3. Prints the new list.

````python
even = [n for n in range(1,21) if n%2 == 0]
print(even)
````

Task: Write a Python program that:
1. Creates a list of student names: ["Ali", "Fatima", "Hassan"].
2. Creates a list of marks: [85, 92, 78].
3. Uses zip() to pair each student with their marks.
4. Uses enumerate(..., start=1) to number each student.
5. Prints each student's number, name, and marks.

````python
student = ["Ali", "Fatima", "Hassan"]
marks = [85, 92, 78]
for rank, (name, mark) in enumerate(zip(student, marks), start=1):
    print(f"{rank}: {name} {mark}")
````
