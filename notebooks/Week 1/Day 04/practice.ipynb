```python
list = []
list.append("jk")
list.append("pk")
list.append("kumar")
removed = list.pop(0)
print(removed)
print(list)

```

```python
value = (5, 8, 5, 9)
print(value.count(5))
print(value.index(8))
print(len(value))
print(sum(value))
```

```python
import numpy as np
array = np.array([10,20,30])
result = array * 3
print(array)
print(result)
```

```python
marks = np.array([45, 70, 38, 90])  # Question 6: calculate and select
print("Average:", marks.mean())
passed = marks >= 50  # Make the True/False selection first
print("Selection:", passed)
print("Passing marks:", marks[passed])
```
