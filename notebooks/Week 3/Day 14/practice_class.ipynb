```python
cars = ["Honda", "Mazda", "Nissan"]
years = [2015, 2016, 2018]



plt.plot(cars, years, marker="s", label="Market Data")
plt.xlabel("Cars Data")
plt.ylabel("Years")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

fruits = ["Apple", "Mango", "Banana"]
sales = [20, 40, 30]

plt.bar(fruits, sales)
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.title("Fruit Sales")
plt.show()
```

```python
phone_hours = [1, 2, 3, 4, 5, 6]

maths_marks = [90, 85, 80, 75, 65, 55]
plt.scatter(phone_hours,maths_marks)
plt.title("Phone Hours vs Maths Marks")
plt.xlabel("Phone Hours")
plt.ylabel("Maths Marks")
plt.show()
```

```python
import pandas as pd

data = pd.DataFrame({
    "phone_hours": [1, 2, 3, 4, 5],
    "marks": [95, 85, 80, 70, 60]
})

r = data["phone_hours"].corr(data["marks"])

print("Correlation:", round(r, 2))
```
